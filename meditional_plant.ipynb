```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import string
import re
```

```python
stopwords = stopwords.words('english')
lemmatizer = WordNetLemmatizer()
```

```python
def clean_text(text):
  text = text.lower()
  text = re.sub(r'[^\w\s+]',' ',text)
  text = re.sub(r'\d+',' ',text)
  text = re.sub(r'\s+',' ',text)
  text = word_tokenize(text)
  text = [word for word in text if word not in stopwords]
  text = [lemmatizer.lemmatize(word) for word in text ]
  return " ".join(text)
```

```python
df['Symptoms'] = df['Symptoms'].apply(clean_text)
```

```python
x = df['Symptoms']
```

```python
from sklearn.feature_extraction.text import CountVectorizer
tfidf = CountVectorizer()
```

```python
vector = tfidf.fit_transform(x)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vector)
```

```python
user = input('Enter a plant name or symptoms:- ')
plant_info = df[df['Plant'].str.lower() == user.lower()]
if len(plant_info) > 0:
  print('For Plant:')
  print(plant_info.iloc[0]['Plant'])
  print('\nUses:')
  print(plant_info.iloc[0]['Uses'])
  print('\nBenefits:')
  print(plant_info.iloc[0]['Benefits'])
else:

  user_vector  = clean_text(user)
  user_vector = tfidf.transform([user_vector])
  best_match = np.argmax(cosine_similarity(user_vector,vector))
  print('Plant Name:')
  print(df.iloc[best_match]['Plant'])
  print('\nBenefits:')
  print(df.iloc[best_match]['Benefits'])
  print('\nHow to Use:')
  print(df.iloc[best_match]['How to Use'])
```
